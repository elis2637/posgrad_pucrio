```python
# CAMADA BRONZE
# Databricks notebook source
# COMMAND ----------
from pyspark.sql.functions import current_timestamp, lit

# 1. Ingestão do Dataset Egressos
df_egressos_bronze_raw = (
    spark.table("mvp_eng_dados.default.dataset_egressos_posgrad")
    .withColumn("_data_ingestao", current_timestamp())
    .withColumn("_arquivo_origem", lit("mvp_eng_dados.default.dataset_egressos_posgrad"))
)

# 2. Ingestão do Dataset Cursos
df_cursos_bronze_raw = (
    spark.table("mvp_eng_dados.default.dataset_cursos")
    .withColumn("_data_ingestao", current_timestamp())
    .withColumn("_arquivo_origem", lit("mvp_eng_dados.default.dataset_cursos"))
)

# Criação do schema bronze se não existir
spark.sql("CREATE SCHEMA IF NOT EXISTS mvp_eng_dados.bronze")

# Gravação nas tabelas Delta Bronze
df_egressos_bronze_raw.write.mode("append").saveAsTable("mvp_eng_dados.bronze.dataset_egressos_bruto")
df_cursos_bronze_raw.write.mode("append").saveAsTable("mvp_eng_dados.bronze.dataset_cursos_bruto")
```
